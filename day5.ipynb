```python
import numpy as np
```

```python
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([0, 0, 0, 1, 1])
```

```python
w = 0.0
b = 0.0
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def forward(X, w, b):
    z = X * w + b
    y_pred = sigmoid(z)
    return z, y_pred
```

```python
def loss_function(y, y_pred):
    epsilon = 1e-8
    
    loss = -np.mean(
        y * np.log(y_pred + epsilon) +
        (1 - y) * np.log(1 - y_pred + epsilon)
    )
    
    return loss
```

```python
z, y_pred = forward(X, w, b)

print("z:", z)
print("Predictions:", y_pred)
```

```python
loss = loss_function(y, y_pred)

print("Loss:", loss)
```

```python
dw = np.mean(X * (y_pred - y))
db = np.mean(y_pred - y)
```

```python
learning_rate = 0.1
w = w - learning_rate * dw
b = b - learning_rate * db
```

```python
epochs = 1000
learning_rate = 0.1

for epoch in range(epochs):

    # Forward propagation
    z, y_pred = forward(X, w, b)

    # Calculate loss
    loss = loss_function(y, y_pred)

    # Calculate gradients
    dw = np.mean(X * (y_pred - y))
    db = np.mean(y_pred - y)

    # Update parameters
    w = w - learning_rate * dw
    b = b - learning_rate * db

    if (epoch + 1) % 100 == 0:
        print(
            f"Epoch {epoch + 1}, Loss: {loss:.4f}"
        )
```

```python
z, y_pred = forward(X, w, b)

print("Final probabilities:")
print(y_pred)
```
